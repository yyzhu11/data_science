## Leetcode 412. Fizz Buzz

Given an integer n, return a string array answer (1-indexed) where:

answer[i] == "FizzBuzz" if i is divisible by 3 and 5.
answer[i] == "Fizz" if i is divisible by 3.
answer[i] == "Buzz" if i is divisible by 5.
answer[i] == i (as a string) if none of the above conditions are true.
 

- Example 1:

Input: n = 3
Output: ["1","2","Fizz"]
- Example 2:

Input: n = 5
Output: ["1","2","Fizz","4","Buzz"]
- Example 3:

Input: n = 15
Output: ["1","2","Fizz","4","Buzz","Fizz","7","8","Fizz","Buzz","11","Fizz","13","14","FizzBuzz"]

````python
def fizzBuzz(n):
    result = []
    for i in range(1, n+1):
        string = ""
        if i % 3 == 0 and i % 5 == 0:
            string = "FizzBuzz"
        elif i % 3 == 0:
            string = "Fizz"
        elif i % 5 == 0:
            string = "Buzz"
        else:
            string = str(i)
        result.append(string)
    return result
````

````python
n = 3
fizzBuzz(n)
````

````python
n = 5
fizzBuzz(n)
````

````python
n = 15
fizzBuzz(n)
````

## Leetcode 537. Complex Number Multiplication (Medium)

A complex number can be represented as a string on the form "real+imaginaryi" where:

real is the real part and is an integer in the range [-100, 100].
imaginary is the imaginary part and is an integer in the range [-100, 100].
$i^2 == -1$.
Given two complex numbers num1 and num2 as strings, return a string of the complex number that represents their multiplications.

- Example 1:

Input: num1 = "1+1i", num2 = "1+1i"
Output: "0+2i"

Explanation: (1 + i) * (1 + i) = 1 + i2 + 2 * i = 2i, and you need convert it to the form of 0+2i.
- Example 2:

Input: num1 = "1+-1i", num2 = "1+-1i"
Output: "0+-2i"

Explanation: (1 - i) * (1 - i) = 1 + i2 - 2 * i = -2i, and you need convert it to the form of 0+-2i.

````python
def complexMult(num1, num2):
    #parse num1, num2
    real1, imag1 = map(int, num1[:-1].split("+"))
    real2, imag2 = map(int, num2[:-1].split("+"))
    result_real = real1*real2 - imag1*imag2
    result_imag = real1*imag2 + real2*imag1
    return f"{result_real}+{result_imag}i"
````

````python
num1 = "1+1i"
num2 = "1+1i"
complexMult(num1, num2)
````

````python
num1 = "1+-1i"
num2 = "1+-1i"
complexMult(num1, num2)
````

## Leetcode 28. Implement strstr(), Find the Index of the First Occurrence in a String

Given two strings needle and haystack, return the index of the first occurrence of needle in haystack, or -1 if needle is not part of haystack.

- Example 1:

Input: haystack = "sadbutsad", needle = "sad"
Output: 0
Explanation: "sad" occurs at index 0 and 6.
The first occurrence is at index 0, so we return 0.
    
- Example 2:

Input: haystack = "leetcode", needle = "leeto"
Output: -1
Explanation: "leeto" did not occur in "leetcode", so we return -1

````python
def findSubstring(s1, s2):
    n1 = len(s2)
    for i in range(len(s1)-n1):
        if s1[i:i+n1] == s2:
            return i
    return -1
````

````python
findSubstring("sadbutsad", "sad")
````

````python
findSubstring("leetcode", "leeto")
````

````python
def strstr2(haystack, needle):
    m = len(needle)
    n = len(haystack)

    if m == 0:
        return 0
    if n == 0:
        return -1
    if m > n:
        return -1
    for i in range(n-m+1):
        if haystack[i:i+m] == needle:
            return i
    return -1
````

````python
#brute force
#1. needle empty, haystack empty, return 0
def strstr(haystack, needle):
    m = len(needle)
    n = len(haystack)

    if m == 0:
        return 0
    if n == 0:
        return -1
    if m > n:
        return -1

    i = j = 0
    start = i
    while i < n and j < m:
        if needle[j] == haystack[i]:
            #print(i, j)
            i += 1
            j += 1
            if (j == m ) and (i < n+1):
                return start
        else:
            i += 1
            j = 0
            start = i
    return -1
````

````python
strstr("leetcode", "leeto")
````

## Leetcode 1003. Check If Word Is Valid After Substitutions (Medium)

Given a string s, determine if it is valid.

A string s is valid if, starting with an empty string t = "", you can transform t into s after performing the following operation any number of times:

Insert string "abc" into any position in t. More formally, t becomes tleft + "abc" + tright, where t == tleft + tright. Note that tleft and tright may be empty.
Return true if s is a valid string, otherwise, return false.

- Example 1:

Input: s = "aabcbc"
Output: true

Explanation:
"" -> "abc" -> "aabcbc"
Thus, "aabcbc" is valid.
- Example 2:

Input: s = "abcabcababcc"
Output: true

Explanation:
"" -> "abc" -> "abcabc" -> "abcabcabc" -> "abcabcababcc"
Thus, "abcabcababcc" is valid.
- Example 3:

Input: s = "abccba"
Output: false

Explanation: It is impossible to get "abccba" using the operation.

````python
def checkWordValid(s):
    if len(s) % 3 != 0:
        return False
    stack = []
    for char in s:
        stack.append(char)
        #print(''.join(stack[-3:]))
        if ''.join(stack[-3:]) == 'abc':
            stack[-3:] = []
    return len(stack) == 0
````

````python
s = "aabcbc"
checkWordValid(s)
````

````python
s = "abcabcababcc"
checkWordValid(s)
````

````python
s = "abccba"
checkWordValid(s)
````

## Leetcode 58. Length of Last Word

Given a string s consisting of words and spaces, return the length of the last word in the string.

A word is a maximal substring consisting of non-space characters only.


- Example 1:

Input: s = "Hello World"
Output: 5
Explanation: The last word is "World" with length 5.
- Example 2:

Input: s = "   fly me   to   the moon  "
Output: 4
Explanation: The last word is "moon" with length 4.
- Example 3:

Input: s = "luffy is still joyboy"
Output: 6
Explanation: The last word is "joyboy" with length 6.

````python
def lengthLastWord(s):
    end = len(s) - 1
    while s[end] == ' ':
        end -= 1
    start = end
    #print(start)
    while s[end]!=' ':
        end -= 1
    #print(end)
    return start - end
    
````

````python
s = "Hello World"
lengthLastWord(s)
````

````python
s = " fly me to the moon "
lengthLastWord(s)
````

````python
s = "luffy is still joyboy"
lengthLastWord(s)
````

## Leetcode 387. First Unique Character in a String

Given a string s, find the first non-repeating character in it and return its index. If it does not exist, return -1.

- Example 1:

Input: s = "leetcode"

Output: 0

Explanation:

The character 'l' at index 0 is the first character that does not occur at any other index.

- Example 2:

Input: s = "loveleetcode"

Output: 2

- Example 3:

Input: s = "aabb"

Output: -1

````python
def firstUniqueChar(s):
    counter = Counter(s)
    for index, char in enumerate(s):
        if counter[char] == 1:
            return index
    return -1    
````

## Leetcode 520. Detect Capital
We define the usage of capitals in a word to be right when one of the following cases holds:

- All letters in this word are capitals, like "USA".
- All letters in this word are not capitals, like "leetcode".
- Only the first letter in this word is capital, like "Google".

Given a string word, return true if the usage of capitals in it is right.

- Example 1:

Input: word = "USA"
Output: true

- Example 2:

Input: word = "FlaG"
Output: false

````python
def detectCapital(word):
    if word.upper() == word:
        return True
    if word.lower() == word:
        return True
    if word.capitalize() == word:
        return True
    return False
````

## Leetcode 159. Longest Substring with At Most Two Distinct Characters (Medium)

Description
Given a string s, return the length of the longest substring that contains at most two distinct characters.

- Example 1:

Input: s = "eceba"
Output: 3
Explanation: The substring is "ece" which its length is 3.

- Example 2:

Input: s = "ccaabbb"
Output: 5
Explanation: The substring is "aabbb" which its length is 5.

````python
def longestTwoDistinct(myString):
    if not myString:
        return 0
    maxLength = 0

    for i in range(len(myString)):
        current = i
        charList = []
        #print('i:', i)
        while current < len(myString):
            #print('current:', current)
            #print('charList:', charList)
            if myString[current] not in charList and len(charList)< 2:
                charList.append(myString[current])
                current += 1
            elif myString[current] not in charList and len(charList) >= 2:
                break
            else:
                current += 1
        length = current - i
        if length > maxLength:
            maxLength = length
    return maxLength
````

````python
longestTwoDistinct("ccaabbb")
````

````python
longestTwoDistinct("eceba")
````

````python
from collections import Counter
def lengthOfLongestSubstringTwoDistinct(s: str) -> int:
        cnt = Counter()
        ans = j = 0
        for i, c in enumerate(s):
            cnt[c] += 1
            while len(cnt) > 2:
                cnt[s[j]] -= 1
                if cnt[s[j]] == 0:
                    cnt.pop(s[j])
                j += 1
            ans = max(ans, i - j + 1)
        return ans
````

## Leetcode 344. Reverse String

Write a function that reverses a string. The input string is given as an array of characters s.

You must do this by modifying the input array in-place with O(1) extra memory.

- Example 1:

Input: s = ["h","e","l","l","o"]
Output: ["o","l","l","e","h"]
- Example 2:

Input: s = ["H","a","n","n","a","h"]
Output: ["h","a","n","n","a","H"]

````python
def reserveString(s):
    start, end = 0, len(s)-1
    while start < end:
        s[start], s[end] = s[end], s[start]
        start += 1
        end -= 1
    return s
````

````python
s = ["h","e","l","l","o"]
reserveString(s)
````

````python
s = ["H","a","n","n","a","h"]
reserveString(s)
````

## Leetcode 541. Reverse String II

Given a string s and an integer k, reverse the first k characters for every 2k characters counting from the start of the string.

If there are fewer than k characters left, reverse all of them. If there are less than 2k but greater than or equal to k characters, then reverse the first k characters and leave the other as original.


- Example 1:

Input: s = "abcdefg", k = 2
Output: "bacdfeg"

- Example 2:

Input: s = "abcd", k = 2
Output: "bacd"

````python
def reverseStringII(s, k):
    slist = list(s)
    start = 0
    n = len(s)
    for start in range(0, n, 2*k):
        end = min(n-1, start+k-1)
        start1 = start        
        while start1 < end:
            slist[start1], slist[end] = slist[end], slist[start1]
            start1 += 1
            end -= 1
        
    return ''.join(slist)
````

````python
s = "abcdefg"
k = 2
reverseStringII(s, k)
````

````python
s = "abcd"
k = 2
reverseStringII(s, k)
````

## Leetcode 151. Reverse Words in a String (Medium)

Given an input string s, reverse the order of the words.

A word is defined as a sequence of non-space characters. The words in s will be separated by at least one space.

Return a string of the words in reverse order concatenated by a single space.

Note that s may contain leading or trailing spaces or multiple spaces between two words. The returned string should only have a single space separating the words. Do not include any extra spaces.

- Example 1:

Input: s = "the sky is blue"
Output: "blue is sky the"

- Example 2:

Input: s = "  hello world  "
Output: "world hello"
Explanation: Your reversed string should not contain leading or trailing spaces.

- Example 3:

Input: s = "a good   example"
Output: "example good a"
Explanation: You need to reduce multiple spaces between two words to a single space in the reversed string.

````python
def reverse_words(myStrings):
    if not myStrings:
        return []
    reversed_list =''
    n = len(myStrings)
    j = n
    for i in range(n-1, -1, -1):
        if myStrings[i] == ' ':
            j = i
        elif (i == 0 or myStrings[i-1] == ' '):
            if len(reversed_list) != 0:
                reversed_list +=' '
            reversed_list += (myStrings[i:j])
    return reversed_list
````

## Leetcode 186. Reverse Words in a String II (Medium)

Given a character array s, reverse the order of the words.

A word is defined as a sequence of non-space characters. The words in s will be separated by a single space.

Your code must solve the problem in-place, i.e. without allocating extra space.

- Example 1:

Input: s = ["t","h","e"," ","s","k","y"," ","i","s"," ","b","l","u","e"]
Output: ["b","l","u","e"," ","i","s"," ","s","k","y"," ","t","h","e"]

- Example 2:

Input: s = ["a"]
Output: ["a"]

````python
def reverse2(slist, start, end):
    #reverse the list
    i = start
    j = end
    while i < j:
        slist[i], slist[j] = slist[j], slist[i]
        i += 1
        j -= 1
    return slist

def reverse_word2(slist):
    reverse2(slist, 0, len(slist)-1)
    j = 0
    for i in range(len(slist)):
        if slist[i] == ' ':
            reverse2(slist, j, i-1)
            j = i + 1
    reverse2(slist, j, i)
    return slist
````
