## Leetcode 136. Single Number

Given a non-empty array of integers, every element appears twice except for one. Find that single one.

Note:
Your algorithm should have a linear runtime complexity. Could you implement it without using extra memory?

- Example: Input: [2,2,1], Output: 1


````python
# use math
def singleNumber(nlist):
    mySet = set(nlist)
    return sum(mySet)*2 - sum(nlist)
````

````python
# use dictionary
def singleNumber2(nlist):
    myDict = {}
    for i in nlist:
        myDict[i] = myDict.get(i, 0) + 1
    for i in myDict.keys():
        if myDict[i] == 1:
            return i
````

````python
# use
def singleNumber3(nlist):
    if not nlist:
        return None
    result = nlist[0]
    for i in nlist[1:]:
        result = result ^ i
    return result
````

````python
singleNumber3([2,2,1])
````

````python
singleNumber([-2,-2,1])
````

## Leetcode 137. Single Number II (Medium)

Given an integer array nums where every element appears three times except for one, which appears exactly once. Find the single element and return it.

You must implement a solution with a linear runtime complexity and use only constant extra space.

- Example 1:

Input: nums = [2,2,3,2]
Output: 3

- Example 2:

Input: nums = [0,1,0,1,0,1,99]
Output: 99

````python
from collections import Counter
def singleNumberII(nums):
    counter = Counter(nums)
    return [num for num in counter.keys() if counter[num] == 1][0]
````

````python
def singleNumberII2(nums):
    result = 0
    for i in range(32):
        bit_count = sum((num >> i) & 1 for num in nums)
        if bit_count % 3:
            if i == 31:
                result -= (1 << i)
            else:
                result |= (1 << i)
    return result          
````

````python
nums = [2,2,3,2]
singleNumberII(nums)
````

````python
nums = [0,1,0,1,0,1,99]
singleNumberII(nums)
````

## Leetcode 169. Majority Element

Description:
Given an array of size n, find the majority element. The majority element is the element that appears more than ⌊ n/2 ⌋ times.

Note:
You may assume that the array is non-empty and the majority element always exist in the array.

- Example 1:

Input: nums = [3,2,3]
Output: 3
- Example 2:

Input: nums = [2,2,1,1,1,2,2]
Output: 2

````python
def majorityElement(nums):
    if not nums:
        return None
    n = len(nums)
    myDict = {}
    for i in nums:
        myDict[i] = myDict.get(i, 0) + 1
    for i in myDict.keys():
        if myDict[i] > n//2:
            return i
````

````python
#Boyer-Moore Voting Algorithm
def majorityElement2(nums):
    candidate = 0
    count = 0
    for num in nums:
        if count == 0:
            candidate = num
            count = 1
        else:
            if candidate == num:
                count += 1
            else:
                count -= 1
    return candidate
````

````python
nums = [3,2,3]
majorityElement(nums)
````

````python
nums = [2,2,1,1,1,2,2]
majorityElement(nums)
````

````python
nums = [3,2,3]
majorityElement2(nums)
````

````python
nums = [2,2,1,1,1,2,2]
majorityElement2(nums)
````

## Leetcode 1. Two Sum

- Description:
Given an array of integers, return indices of the two numbers such that they add up to a specific target.

Note:
You may assume that each input would have exactly one solution, and you may not use the same element twice.

- Example 1:

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
- Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
- Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]

````python
def twoSum(nums, target):
    myDict = {}
    if not nums:
        return None
    n = len(nums)
    for index in range(n):
        num = nums[index]
        if (target - num) in myDict:
            return [myDict[target - num], index]
        else:
            myDict[num] = index 
    return None  
````

````python
nums = [2,7,11,15]
target = 9
twoSum(nums, target)
````

````python
nums = [3,2,4]
target = 6
twoSum(nums, target)
````

````python
nums = [3,3]
target = 6
twoSum(nums, target)
````

## Leetcode 167. Two Sum II - Input array is sorted (Medium)

Given an array of integers that is already sorted in ascending order,
find two numbers such that they add up to a specific target number.

The function twoSum should return indices of the two numbers such that they add up to the target,
where index1 must be less than index2.

Note:
Your returned answers (both index1 and index2) are not zero-based.
You may assume that each input would have exactly one solution and you may not use the same element twice.

- Example 1:

Input: numbers = [2,7,11,15], target = 9
Output: [1,2]

Explanation: The sum of 2 and 7 is 9. Therefore, index1 = 1, index2 = 2. We return [1, 2].
- Example 2:

Input: numbers = [2,3,4], target = 6
Output: [1,3]

Explanation: The sum of 2 and 4 is 6. Therefore index1 = 1, index2 = 3. We return [1, 3].
- Example 3:

Input: numbers = [-1,0], target = -1
Output: [1,2]

Explanation: The sum of -1 and 0 is -1. Therefore index1 = 1, index2 = 2. We return [1, 2].

````python
#use Two pointers
def twoSum2(nums, target):
    start = 0
    end = len(nums) - 1
    while start < end:
        if nums[start] + nums[end] == target:
            return [start+1, end+1]
        if nums[start] + nums[end] < target:
            start += 1
        else:
            end -= 1
    return None
````

````python
# binary search
def binarySearch(nlist, target, start):
    left = start
    right = len(nlist) - 1
    while left <= right:
        mid = (left + right) // 2
        if nlist[mid] == target:
            return mid
        elif nlist[mid] > target:
            right = mid - 1
        else:
            left = mid + 1
    return None
        
def twoSumBS(nlist, target):
    for index in range(len(nlist)):
        target_index = binarySearch(nlist, target - nlist[index], index+1)
        if target_index is not None:
            return [index+1, target_index+1]
    return []
````

````python
numbers = [2,7,11,15]
target = 9
twoSum2(numbers, target)
````

````python
numbers = [2,3,4]
target = 6
twoSum2(numbers, target)
````

````python
numbers = [-1,0]
target = -1
twoSum2(numbers, target)
````

- Description:

Note:
Your function should return true if any value appears at least twice in the array, and it should return false
if every element is distinct.

- Example 1:

Input: nums = [1,2,3,1]

Output: true

Explanation:

The element 1 occurs at the indices 0 and 3.

- Example 2:

Input: nums = [1,2,3,4]

Output: false

Explanation:

All elements are distinct.

- Example 3:

Input: nums = [1,1,1,3,3,4,3,2,4,2]

Output: true

````python
def containDuplicate(nums):
  if not nums:
    return False
  myDict = {}
  for i in nums:
    if i in myDict:
      return True
    else:
      myDict[i] = myDict.get(i, 0)+1
  return False
````

## Leetcode 219. Contains Duplicate II

- Description:
Given an array of integers and an integer k,find out whether there are two distinct indices i and j
in the array such that nums[i] = nums[j] and the absolute difference between i and j is at most k.

- Example1:
Input: nums = [1,2,3,1], k = 3, Output: true

- Example2:
Input: nums = [1,0,1,1], k = 1, Output: true

- Example3:
Input: nums = [1,2,3,1,2,3], k = 2, Output: false


````python
def containDup(nlist, k):
    if not nlist:
        return False
    myDict = {}
    for index in range(len(nlist)):
        if nlist[index] in myDict and abs(index - myDict[nlist[index]]) <= k :
            return True
        myDict[nlist[index]] = index
    return False
    
````

````python
containDup([1,2,3,1], 3)
````

````python
containDup([1,0,1,1], 1)
````

````python
containDup([1,2,3,1,2,3], 2)
````

## Leetcode 242. Valid Anagram

- Description:
Given two strings s and t , write a function to determine if t is an anagram of s.

Note: You may assume the string contains only lowercase alphabets.

- Example1:
Input: s = "anagram", t = "nagaram", Output: true

- Example2:
Input: s = "rat", t = "car", Output: false


````python
def validAnagram(s, t):
    myDict1 = {}
    myDict2 = {}
    for i in range(len(s)):
        myDict1[s[i]] = myDict1.get(s[i], 0) + 1
    for i in range(len(t)):
        myDict2[t[i]] = myDict2.get(t[i], 0) + 1
    if myDict1 == myDict2:
        return True
    else:
        return False
````

````python
validAnagram("anagram", "nagaram")
````

````python
validAnagram("rat", "car")
````

## Leetcode 268. Missing Number

- Description:
Given an array containing n distinct numbers taken from 0, 1, 2, ..., n,
find the one that is missing from the array.

Note:
Your algorithm should run in linear runtime complexity. Could you implement it using only constant extra space complexity?

- Example1:
Input: [3,0,1], Output: 2

- Example2:
Input: [9,6,4,2,3,5,7,0,1], Output: 8


````python
# math
def missingNum(nlist):
    if not nlist:
        return None
    n = len(nlist) + 1
    return (n)*(n-1)//2 - sum(nlist)
    
````

````python
missingNum([3,0,1])
````

````python
missingNum([9,6,4,2,3,5,7,0,1])
````

````python
# dictionary
def missingNum2(nlist):
    if not nlist:
        return None
    
    myDict = {}
    for i in nlist:
        myDict[i] = myDict.get(i, 0) + 1
    for i in range(len(nlist)+1):
        if i not in myDict:
            return i
````

````python
missingNum2([3,0,1])
````

````python
missingNum([9,6,4,2,3,5,7,0,1])
````

````python
intersectionTwoArray([1,2,2,1], [2,2])
````

````python
def intersectionTwoArray2(nlist1, nlist2):
    myDict = {}
    nlist = []
    for i in nlist1:
        myDict[i] = myDict.get(i, 0) + 1
    for j in nlist2:
        if j in myDict:
            nlist.append(j)
            del myDict[j]
    return nlist
````

## Leetcode 350. Intersection of Two Arrays II

- Description:
Given two arrays, write a function to compute their intersection.

Note:
Each element in the result should appear as many times as it shows in both arrays.
The result can be in any order.

- Example1:
Input: nums1 = [1,2,2,1], nums2 = [2,2], Output: [2,2]

- Example2:
Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4], Output: [4,9]

````python
def intersectionTwoArrayII(nlist1, nlist2):
    myDict = {}
    nlist = []
    for i in nlist1:
        myDict[i] = myDict.get(i, 0) + 1
    for j in nlist2:
        if j in myDict:
            nlist.append(j)
            myDict[j] -= 1
            if myDict[j] == 0:
                del myDict[j]
    return nlist
````

````python
intersectionTwoArrayII([1,2,2,1], [2,2])
````

````python
intersectionTwoArrayII([4,9,5], [9,4,9,8,4])
````

## Leetcode 389. Find the Difference

- Description:
Given two strings s and t which consist of only lowercase letters.String t is generated by random shuffling string s and
then add one more letter at a random position.Find the letter that was added in t.

- Example 1:

s = "abcd"   t = "abcde", Output: e,                 Explanation: "e" is the letter that was added.

s = "abba" t="abbba", output: b



````python
def findDifference(s, t):
    myDict = {}
    for i in range(len(s)):
        myDict[s[i]] = myDict.get(s[i], 0) + 1
    for j in range(len(t)):
        if t[j] in myDict:
            myDict[t[j]] -= 1
            if myDict[t[j]] == 0:
                del myDict[t[j]]
        else:
            return t[j]
    return myDict.keys()
````

````python
findDifference("abcd", "abcde")
````

````python
findDifference("abba", "abbba")
````

## Leetcode 697. Degree of an Array
Given a non-empty array of non-negative integers nums, the degree of this array is defined as the maximum frequency of any one of its elements.

Your task is to find the smallest possible length of a (contiguous) subarray of nums, that has the same degree as nums.

- Example 1:

Input: nums = [1,2,2,3,1]
Output: 2

Explanation: 
The input array has a degree of 2 because both elements 1 and 2 appear twice.
Of the subarrays that have the same degree:
[1, 2, 2, 3, 1], [1, 2, 2, 3], [2, 2, 3, 1], [1, 2, 2], [2, 2, 3], [2, 2]
The shortest length is 2. So return 2.

- Example 2:

Input: nums = [1,2,2,3,1,4,2]
Output: 6

Explanation: 
The degree is 3 because the element 2 is repeated 3 times.
So [2,2,3,1,4,2] is the shortest subarray, therefore returning 6.

````python
import numpy as np
def degreeArray(nums):
    counter = Counter(nums)
    max_frequency = counter.most_common()[0][1]
    # Track first and last occurrence index for each number
    first_occurrence = {}
    last_occurrence = {}
    for index, value in enumerate(nums):
        if value not in first_occurrence:
            first_occurrence[value] = index
        last_occurrence[value] = index
    min_length = np.inf
    for num in nums:
        if counter[num] == max_frequency:
            length = last_occurrence[num] - first_occurrence[num]+1
            min_length= min(min_length, length)
    return min_length
````

## Leetcode 594. Longest Harmonious Subsequence
We define a harmonious array as an array where the difference between its maximum value and its minimum value is exactly 1.

Given an integer array nums, return the length of its longest harmonious subsequence among all its possible subsequences.

- Example 1:

Input: nums = [1,3,2,2,5,2,3,7]

Output: 5

Explanation:

The longest harmonious subsequence is [3,2,2,2,3].

- Example 2:

Input: nums = [1,2,3,4]

Output: 2

Explanation:

The longest harmonious subsequences are [1,2], [2,3], and [3,4], all of which have a length of 2.

- Example 3:

Input: nums = [1,1,1,1]

Output: 0

Explanation:

No harmonic subsequence exists.

````python
def longestHarmoniousSub(nums):
    counter = Counter(nums)
    max_length = 0
    for num, val in counter.items():
        if (num+1) in counter:
            max_length = max(max_length, counter[num]+counter[num+1])
    return max_length
````

````python
nums = [1,3,2,2,5,2,3,7]
longestHarmoniousSub(nums)
````

````python
nums = [1,2,3,4]
longestHarmoniousSub(nums)
````

````python
nums = [1,1,1,1]
longestHarmoniousSub(nums)
````

## Leetcode 442. Find All Duplicates in an Array (Medium)

- Description:
Given an array of integers, 1 ≤ a[i] ≤ n (n = size of array), some elements appear twice and others appear once.Find all the elements that appear twice in this array.

Could you do it without extra space and in O(n) runtime?


- Example:

Input: [4,3,2,7,8,2,3,1], output [2,3]

## Leetcode 767. Reorganize String (Medium)

Given a string s, rearrange the characters of s so that any two adjacent characters are not the same.
Return any possible rearrangement of s or return "" if not possible.

- Example 1:

Input: s = "aab"
Output: "aba"
- Example 2:

Input: s = "aaab"
Output: ""

````python
from collections import Counter
def reorganzieString(s):
    myDict = Counter(s)
    maxCount = max(myDict.values())
    s_len = len(s)    
    if (maxCount -1) > (s_len-maxCount):
        return ""
    result = [None]* s_len
    index = 0
    for char, frequency in myDict.most_common():
        while frequency > 0:
            result[index] = char
            frequency -= 1
            index +=2
            if index >= s_len:
                index = 1
    return ''.join(result)
````

## Leetcode 128. Longest Consecutive Sequence (Medium)

Given an unsorted array of integers nums, return the length of the longest consecutive elements sequence.

You must write an algorithm that runs in O(n) time.

- Example 1:

Input: nums = [100,4,200,1,3,2]
Output: 4
Explanation: The longest consecutive elements sequence is [1, 2, 3, 4]. Therefore its length is 4.
- Example 2:

Input: nums = [0,3,7,2,5,8,4,6,0,1]
Output: 9
- Example 3:

Input: nums = [1,0,1,2]
Output: 3

````python
def longestConsecutiveSeq(nums):
    max_length = 0
    nums_set = set(nums)
    seq_length = {}
    for num in nums:
        if num not in nums_set:
            continue
        seq_end = num
        while seq_end in nums_set:
            nums_set.remove(seq_end)
            seq_end += 1
        length = seq_end - num + seq_length.get(seq_end, 0)
        seq_length[num] = length
        max_length = max(max_length, length)
    return max_length
````

````python
nums = [100,4,200,1,3,2] 
longestConsecutiveSeq(nums)
````

````python
nums = [0,3,7,2,5,8,4,6,0,1]
longestConsecutiveSeq(nums)
````

## Leetcode 621. Task Scheduler (Medium)

You are given an array of CPU tasks, each labeled with a letter from A to Z, and a number n. Each CPU interval can be idle or allow the completion of one task. Tasks can be completed in any order, but there's a constraint: there has to be a gap of at least n intervals between two tasks with the same label.

Return the minimum number of CPU intervals required to complete all tasks.

- Example 1:

Input: tasks = ["A","A","A","B","B","B"], n = 2

Output: 8

Explanation: A possible sequence is: A -> B -> idle -> A -> B -> idle -> A -> B.

After completing task A, you must wait two intervals before doing A again. The same applies to task B. In the 3rd interval, neither A nor B can be done, so you idle. By the 4th interval, you can do A again as 2 intervals have passed.

- Example 2:

Input: tasks = ["A","C","A","B","D","B"], n = 1

Output: 6

Explanation: A possible sequence is: A -> B -> C -> D -> A -> B.

With a cooling interval of 1, you can repeat a task after just one other task.

- Example 3:

Input: tasks = ["A","A","A", "B","B","B"], n = 3

Output: 10

Explanation: A possible sequence is: A -> B -> idle -> idle -> A -> B -> idle -> idle -> A -> B.

There are only two types of tasks, A and B, which need to be separated by 3 intervals. This leads to idling twice between repetitions of these tasks.

````python
def taskScheduler(tasks, n):
    counter= Counter(tasks)
    max_freq = max(count for count in counter.values())
    max_count = sum(1 for count in counter.values() if count == max_freq)

    return max(len(tasks), (max_freq-1)*(n+1)+ max_count)
````

````python
tasks = ["A","A","A","B","B","B"]
n = 2
taskScheduler(tasks, n)
````

````python
tasks = ["A","C","A","B","D","B"]
n = 1
taskScheduler(tasks, n)
````

````python
tasks = ["A","A","A", "B","B","B"]
n = 3
taskScheduler(tasks, n)
````

## Leetcode 49. Group Anagrams (Medium)

Given an array of strings strs, group the anagrams together. You can return the answer in any order.

- Example 1:

Input: strs = ["eat","tea","tan","ate","nat","bat"]

Output: [["bat"],["nat","tan"],["ate","eat","tea"]]

Explanation:

There is no string in strs that can be rearranged to form "bat".
The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.
- Example 2:

Input: strs = [""]

Output: [[""]]

Example 3:

Input: strs = ["a"]

Output: [["a"]]

````python
from collections import defaultdict
def groupAnagrams(strs):
    result = []
    myDict = defaultdict(list)
    for string in strs:
        sorted_string = ''.join(sorted(string))
        myDict[sorted_string].append(string)
    return [value for key, value in myDict.items()]
````

## Leetcode 3. Longest Substring Without Repeating Characters (Medium)

Given a string s, find the length of the longest substring without duplicate characters.

- Example 1:

Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with the length of 3. Note that "bca" and "cab" are also correct answers.

- Example 2:

Input: s = "bbbbb"
Output: 1
Explanation: The answer is "b", with the length of 1.

- Example 3:

Input: s = "pwwkew"
Output: 3
Explanation: The answer is "wke", with the length of 3.
Notice that the answer must be a substring, "pwke" is a subsequence and not a substring.

````python
def longestSubstring(myStr):
    j=0
    result = 0
    myDict = {}
    for i, c in enumerate(myStr):
        if c in myDict:            
            myDict[c] = 1
            while c in myDict:
                myDict[myStr[j]] -= 1
                if myDict[myStr[j]] == 0:
                    del myDict[myStr[j]]
                j += 1
        myDict[c] = 1
        result = max(result, i-j+1)
    return result
````

````python
longestSubstring("abcabcbb")
````

````python
longestSubstring("bbbbb")
````

````python
longestSubstring("pwwkew")
````

## Leetcode 560. Subarray Sum Equals K (Medium)

Given an array of integers nums and an integer k, return the total number of subarrays whose sum equals to k.

A subarray is a contiguous non-empty sequence of elements within an array.

- Example 1:

Input: nums = [1,1,1], k = 2
Output: 2
- Example 2:

Input: nums = [1,2,3], k = 3
Output: 2

````python
def subSumEqualK(nums, k):
    sum_counter = Counter({0:1})
    result = 0
    currenct_sum = 0
    for num in nums:
        currenct_sum += num
        result += sum_counter[currenct_sum-k]
        sum_counter[currenct_sum] += 1
    return result
````

````python
nums = [1,1,1]
k=2
subSumEqualK(nums, k)
````

````python
nums =  [1,2,3]
k=3
subSumEqualK(nums, k)
````

## Leetcode 566. Reshape the Matrix

In MATLAB, there is a handy function called reshape which can reshape an m x n matrix into a new one with a different size r x c keeping its original data.

You are given an m x n matrix mat and two integers r and c representing the number of rows and the number of columns of the wanted reshaped matrix.

The reshaped matrix should be filled with all the elements of the original matrix in the same row-traversing order as they were.

If the reshape operation with given parameters is possible and legal, output the new reshaped matrix; Otherwise, output the original matrix.

- Example 1:

Input: mat = [[1,2],[3,4]], r = 1, c = 4
Output: [[1,2,3,4]]

- Example 2:

Input: mat = [[1,2],[3,4]], r = 2, c = 4
Output: [[1,2],[3,4]]

````python
def reshapeMatrix(mat, r, c):
    n = len(mat) # row
    m = len(mat[0]) #column
    #print(n, m)
    if r*c != m *n:
        return mat
    new_mat =  [[0 for _ in range(c)] for _ in range(r)]
    result = []
    for i in range(n):
        for j in range(m):
            #print(i, j)
            result.append(mat[i][j])
       
    for i in range(r):
        for j in range(c):
            #print(i,j)
            new_mat[i][j] = result[c*i+j]
    return new_mat
    return
````

````python
reshapeMatrix([[1,2],[3,4]], 1, 4)
````

````python
reshapeMatrix([[1,2,3,4]], 2, 2)
````

````python
import math
def maxProductThreeNumbers(nums):
    n = len(nums)
    if n < 3:
        return None
    max_product = math.prod(nums[:3])
    index = 1
    while (index+2) < n:
        new_product = math.prod(nums[index:index+3]) 
        if new_product > max_product:
            max_product = new_product
        index += 1
            
    return max_product    
````

## Leetcode 163. Missing Ranges (Medium)

Description
You are given an inclusive range [lower, upper] and a sorted unique integer array nums, where all elements are within the inclusive range.

A number x is considered missing if x is in the range [lower, upper] and x is not in nums.

Return the shortest sorted list of ranges that exactly covers all the missing numbers. That is, no element of nums is included in any of the ranges, and each missing number is covered by one of the ranges.

- Example 1:

Input: nums = [0,1,3,50,75], lower = 0, upper = 99
Output: [[2,2],[4,49],[51,74],[76,99]]

Explanation: The ranges are:
[2,2]
[4,49]
[51,74]
[76,99]

- Example 2:

Input: nums = [-1], lower = -1, upper = -1
Output: []

Explanation: There are no missing ranges since there are no missing numbers.

````python
def missingRange(nlist, lower, upper):
    i = lower
    j = upper
    n = len(nlist)
    result = []
    if n == 0:
        return [[lower, upper]]
    if lower < nlist[0]:
        result.append([lower, nlist[0] -1])
    k = 1
    while k <  n :
        if (nlist[k-1] + 1 ) < nlist[k]:
            result.append([nlist[k-1] + 1, nlist[k] - 1])
        k = k + 1
    
    if upper > nlist[n-1]:
        result.append([nlist[n-1]+1, upper])
    return result
````

````python
def findSubstring(s1, s2):
    n1 = len(s2)
    for i in range(len(s1)-n1):
        if s1[i:i+n1] == s2:
            return i
    return -1
````

````python
def strstr2(haystack, needle):
    m = len(needle)
    n = len(haystack)
       
    if m == 0:
        return 0
    if n == 0:
        return -1
    if m > n:
        return -1
    for i in range(n-m+1):
        if haystack[i:i+m] == needle:
            return i
    return -1
````

````python
#brute force
#1. needle empty, haystack empty, return 0
def strstr(haystack, needle):
    m = len(needle)
    n = len(haystack)
       
    if m == 0:
        return 0
    if n == 0:
        return -1
    if m > n:
        return -1

    i = j = 0
    start = i
    while i < n and j < m:
        if needle[j] == haystack[i]:
            #print(i, j)
            i += 1
            j += 1
            if (j == m ) and (i < n+1):
                return start
        else:
            i += 1
            j = 0
            start = i
    return -1
            
````

## Leetcode 159. Longest Substring with At Most Two Distinct Characters (Medium)

Description
Given a string s, return the length of the longest substring that contains at most two distinct characters.

- Example 1:

Input: s = "eceba"
Output: 3
Explanation: The substring is "ece" which its length is 3.

- Example 2:

Input: s = "ccaabbb"
Output: 5
Explanation: The substring is "aabbb" which its length is 5.

````python
def longestTwoDistinct(myString):
    if not myString:
        return 0
    maxLength = 0

    for i in range(len(myString)):
        current = i
        charList = []
        #print('i:', i)
        while current < len(myString):  
            #print('current:', current)
            #print('charList:', charList)
            if myString[current] not in charList and len(charList)< 2:
                charList.append(myString[current])
                current += 1
            elif myString[current] not in charList and len(charList) >= 2:
                break
            else:
                current += 1
        length = current - i
        if length > maxLength:
            maxLength = length
    return maxLength    
            
````

````python
longestTwoDistinct("eceba")
````

````python
from collections import Counter
def lengthOfLongestSubstringTwoDistinct(s: str) -> int:
        cnt = Counter()
        ans = j = 0
        for i, c in enumerate(s):
            cnt[c] += 1
            while len(cnt) > 2:
                cnt[s[j]] -= 1
                if cnt[s[j]] == 0:
                    cnt.pop(s[j])
                j += 1
            ans = max(ans, i - j + 1)
        return ans
````

A word is defined as a sequence of non-space characters. The words in s will be separated by at least one space.

Return a string of the words in reverse order concatenated by a single space.

Note that s may contain leading or trailing spaces or multiple spaces between two words. The returned string should only have a single space separating the words. Do not include any extra spaces.

- Example 1:

Input: s = "the sky is blue"
Output: "blue is sky the"

- Example 2:

Input: s = "  hello world  "
Output: "world hello"
Explanation: Your reversed string should not contain leading or trailing spaces.

- Example 3:

Input: s = "a good   example"
Output: "example good a"
Explanation: You need to reduce multiple spaces between two words to a single space in the reversed string.

````python
def reverse_words(myStrings):
    if not myStrings:
        return []
    reversed_list =''
    n = len(myStrings)
    j = n
    for i in range(n-1, -1, -1):
        if myStrings[i] == ' ':
            j = i
        elif (i == 0 or myStrings[i-1] == ' '):
            if len(reversed_list) != 0:
                reversed_list +=' '
            reversed_list += (myStrings[i:j])
    return reversed_list   
````

Given a character array s, reverse the order of the words.

Your code must solve the problem in-place, i.e. without allocating extra space.

- Example 1:

Input: s = ["t","h","e"," ","s","k","y"," ","i","s"," ","b","l","u","e"]
Output: ["b","l","u","e"," ","i","s"," ","s","k","y"," ","t","h","e"]

- Example 2:

Output: ["a"]

````python
def reverse2(slist, start, end):
    #reverse the list
    i = start
    j = end
    while i < j:
        slist[i], slist[j] = slist[j], slist[i]
        i += 1
        j -= 1
    return slist

def reverse_word2(slist):
    reverse2(slist, 0, len(slist)-1)
    j = 0
    for i in range(len(slist)):
        if slist[i] == ' ':
            reverse2(slist, j, i-1)
            j = i + 1
    reverse2(slist, j, i)
    return slist
````
